```python
#导入数据
df = pd.read_csv(r'C:\Users\ASUS\Project\Project\data\猫眼电影.csv')
```

```python
#数据清洗
df['country'] =  (df['pub_time'].str.split('(').str[1]).str.split(')').str[0]
df['pub_time'] = df['pub_time'].str.split('(').str[0]
df = df[df['pub_time'].str.len()>6]
df = df.dropna()
```

```python
#数据重构
df['pub_time'] = pd.to_datetime(df['pub_time'])
df['year'] = df['pub_time'].dt.year
```

```python
#数据分箱
labels = ['低时长','中时长','高时长','超长']
df['long_time_level'] = pd.cut(df['pub_time'], bins=4, labels=labels)
```

```python
a = ','.join([df['author'].iloc[i] for i in range(0,df.shape[0])])
b = a.split(',')

author_count = Counter(b)
print(author_count)
# 4. 转成 DataFrame 方便查看（按次数从多到少排序）
result = pd.DataFrame(author_count.items(), columns=['作者', '出现次数'])
result = result.sort_values('出现次数', ascending=False)
print(result)
```
